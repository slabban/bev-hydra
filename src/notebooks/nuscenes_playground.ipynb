```python
from nuscenes.nuscenes import NuScenes
```

```python
data_root = '/home/slabban/workspace/bev-hydra/data/nuscenes/trainval'
```

```python
nusc = NuScenes(version='v1.0-trainval', dataroot=data_root, verbose=True)
```

Get first sample and print it's contents

```python
sample = nusc.sample[0]
print("Sample Info:")
print(sample)
```

Go through categories in sample token

```python
for annotation_token in sample['anns']:
    # Filter out all non vehicle instances
    annotation = nusc.get('sample_annotation', annotation_token)
    print(annotation['category_name'])
```
